# Part 2. 바이브 코딩이 뭐길래?

## 전통적인 코딩 vs 바이브 코딩

| | 전통적 | 바이브 코딩 |
|---|---|---|
| 무엇을 하나 | 문법 외우고 직접 타이핑 | AI에게 한국어로 부탁 |
| 배우는 데 걸리는 시간 | 수개월~수년 | 30분 |
| 막혔을 때 | 구글링, 책 | "이거 안 돼" 다시 부탁 |
| 한계 | 본인 실력만큼 | AI 실력만큼 (꽤 높음) |

## 바이브 코딩이 잘 되는 영역과 아닌 영역

**✅ 잘 됨**
- 데이터 분석, 시각화
- 머신러닝 모델 만들기
- 간단한 웹 페이지
- 일상적 자동화 스크립트

**⚠️ 주의**
- 실시간 시스템, 안전 critical (의료기기 등)
- 완전 신규 분야 (AI도 모름)
- 보안 코드 (취약점 검증 필요)

## 자주 망하는 패턴

### 🚨 1. 셀 실행 순서 어긋남
Colab은 위에서 아래로 셀을 실행한다고 가정합니다. 중간 셀을 건너뛰거나 거꾸로 실행하면 변수가 없다고 에러가 납니다.

**해결**: 메뉴 → 런타임 → **모두 실행** (Ctrl+F9)

### 🚨 2. AI가 잘못된 라이브러리 import
AI가 가끔 존재하지 않는 라이브러리를 import하라고 합니다. 그러면 에러나죠.

**해결**: 에러 메시지 그대로 보여주고 "다시 짜줘"

프롬프트에 데이터 모양을 명시하지 않으면 AI가 *추측*합니다. 가끔 맞고 가끔 틀립니다.

**해결**: 프롬프트에 컬럼명, dtype, shape 명시. 또는 `df.head()` 결과를 같이 붙여서 부탁.
